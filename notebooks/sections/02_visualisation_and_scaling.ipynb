<h3>5.1.1 Analysis Using Histogram</h3>

<p>I will be utilising Histograms to visualise the distributions of my features. It will show me the frequency of the features in my dataset. <h1>5. Scaling of Features</h1>

<p>As most clustering models requires Normalisation as different features have different unit of measurements and ranges, I will be scaling my Features to ensure large numerical values do not dominate the clustering and leave the numerical values of lower counts redundant. There is 2 main methods for Scaling Features for which I will be utilising. These two methods are listed below in the sub-sections.</p>

<hr/>
<h2>5.1 Z-Score Standardisation</h2>

<p>Z-Score Standardisation will scale down the data such that the mean will be 0 while the Standard Deviation is 1 by calculating the Z-Score of the dataset. The formula is listed below as reference to how this works.</p>

$$
X' = \frac{X - \mu}{\sigma}
$$

**Where:**
- $X'$ is the standardised feature value
- $X$ is the original feature value
- $\mu$ is the mean of the feature
- $\sigma$ is the standard deviation of the feature

<br/>
<p>With reference to the purpose and capability of the Z-Score Standardisation, I will proceed to scale the data using the Z-Score Standardisation in the cell below.</p>

```python
# ----------- Selecting Numerical Features For Scaling ----------- #
features_to_scale = ['Age', 'Income (k$)', 'How Much They Spend']

# ------------ Applying Z-Score Standardisation ----------- #
scaler_zscore = StandardScaler()
df_zscore = df.copy()
df_zscore[features_to_scale] = scaler_zscore.fit_transform(df[features_to_scale])

# ------------ Displaying Z-Score Standardisation DataFrame ----------- #
df_zscore.head(5).style.background_gradient(cmap="Blues")
```

<p>With reference to the output of the DataFrame above, I am able to confirm that the Z-Score Standardisation have been executed successfully as the Data are now scaled accordingly for all Numerical Data Column.</p>

<hr/>
<h2>5.2 Min-Max Scaler</h2>

<p>Min-Max Scaler will scale the data down such that it ranges between 0 and 1.</p>

$$
X' = \frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

**Where:**
- $X'$ is the scaled feature value
- $X$ is the original feature value
- $X_{\min}$ is the minimum value of the feature
- $X_{\max}$ is the maximum value of the feature

<p>With the function fo the Min-Max Scaler indicated, I will proceed to utilise the Min-Max Scaler to scale my data in the cell below.</p>

```python
# ----------- Selecting Numerical Features For Scaling ----------- #
features_to_scale = ['Age', 'Income (k$)', 'How Much They Spend']

# ------------ Applying Min-Max Scaling ----------- #
scaler_minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[features_to_scale] = scaler_minmax.fit_transform(df[features_to_scale])

# ------------ Displaying Z-Score Standardisation DataFrame ----------- #
df_minmax.head(5).style.background_gradient(cmap="Blues")
```

<p>With reference to the output of the DataFrame above, I am able to confirm that the Min-Max Scaler have been executed successfully as the Data are now scaled accordingly for all Numerical Data Column.</p>

<hr/>
<h2>5.3 Visualising Scaled Data</h2>

<p>In order to verify that the integrity of the Data is still the same as the Original Data despite the scaling, I will be plotting 3 sets of Histograms that represents the Original Data, Z-Score Standardised Data and Min-Max Scaled Data. I will them compare the kde for each of the plots in the same Data Column so as to ensure that the kde is generally similar but the value isn't due to scaling. The Histogram will be plot in the cell below.</p>

```python
# ----------- Seaborn Styling ----------- #
sns.set_style("whitegrid")

# ------------ Plotting Of Histogram ----------- #
def compare_scaling(feature):
    plt.figure(figsize=(12, 5))

    # Original Data #
    plt.subplot(1, 3, 1)
    sns.histplot(df[feature], bins=20, kde=True, color="blue", alpha=0.2, edgecolor="black")
    plt.title(f'Original {feature}', fontsize=12)
    plt.xlabel(feature)

    # Z-Score Standardisation #
    plt.subplot(1, 3, 2)
    sns.histplot(df_zscore[feature], bins=20, kde=True, color="green", alpha=0.2, edgecolor="black")
    plt.title(f'Z-Score Standardised {feature}', fontsize=12)
    plt.xlabel(feature)

    # Min-Max Scaling #
    plt.subplot(1, 3, 3)
    sns.histplot(df_minmax[feature], bins=20, kde=True, color="red", alpha=0.2, edgecolor="black")
    plt.title(f'Min-Max Scaled {feature}', fontsize=12)
    plt.xlabel(feature)

    # Displaying Plot #
    plt.tight_layout()
    plt.show()

# ------------ Plotting For All Features ----------- #
for feature in features_to_scale:
    compare_scaling(feature)
```

<p>With reference to the output of the Histograms with the KDEs, I am able to verify that the Scaling did not affect the integrity or the distribution of the data, instead, the only thing that has been affected is the Numerical Data which was all scalled to a lower amount so as to ensure Normalisation amongst all the Numerical Data in the Dataset.</p>
